```python
import pandas as pd
import joblib
import shap
```

```python
batsman_df = pd.read_csv("Datasets/Final_Features/model_dataset.csv")

TARGET = "target_next_match_runs"
X_bat = batsman_df.drop(columns=[TARGET])
```

```python
#Trained Model
batsman_model = joblib.load("xgb_model.joblib")
```

```python
# Create SHAP Explainer
explainer_bat = shap.TreeExplainer(batsman_model)
# Calculate SHAP Values
shap_values_bat = explainer_bat.shap_values(X_bat)
# Summary Plot
shap.summary_plot(shap_values_bat, X_bat)
```

```python
#Bowler
bowler_df = pd.read_csv("Datasets/Final_Features/final_bowler_dataset.csv")

TARGET = "target_next_match_wickets"
X_bowl = bowler_df.drop(columns=[TARGET])
```

```python
# Trained Model
bowler_model = joblib.load("bowler_rfmodel.joblib")
```

```python
# Create SHAP Explainer
explainer_bowl = shap.TreeExplainer(bowler_model)
# Calculate SHAP Values
shap_values_bowl = explainer_bowl.shap_values(X_bowl)
# Summary Plot
shap.summary_plot(shap_values_bowl, X_bowl)
```
