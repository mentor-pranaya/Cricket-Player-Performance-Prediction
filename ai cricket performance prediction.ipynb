```python
# Fill numeric missing values with 0
num_cols = ball_by_ball.select_dtypes(include='number').columns
ball_by_ball[num_cols] = ball_by_ball[num_cols].fillna(0)

# Fill categorical missing values
cat_cols = ball_by_ball.select_dtypes(include='object').columns
ball_by_ball[cat_cols] = ball_by_ball[cat_cols].fillna("Unknown")
```

```python
num_cols = deliveries.select_dtypes(include='number').columns
deliveries[num_cols] = deliveries[num_cols].fillna(0)

cat_cols = deliveries.select_dtypes(include='object').columns
deliveries[cat_cols] = deliveries[cat_cols].fillna("Unknown")
```

```python
matches = matches.fillna("Unknown")
```

```python
def normalize_season(season):
    season = str(season)
    return int(season.split('/')[1]) if '/' in season else int(season)

matches['season'] = matches['season'].apply(normalize_season)

sorted(matches['season'].unique())
```

```python
ball_by_ball['total_runs'] = (
    ball_by_ball['batsman_runs'] + ball_by_ball['extras']
)
```

```python
deliveries.drop_duplicates(inplace=True)
matches.drop_duplicates(inplace=True)
ball_by_ball.drop_duplicates(inplace=True)
```

```python
ball_by_ball = ball_by_ball[
    ['matchId','batsman','bowler','batting_team','bowling_team',
     'total_runs']
]
```

```python
matches = matches[['matchId','season','venue']]
```

```python
player_match = (
    ball_by_ball.groupby(
        ['matchId','batsman','batting_team','bowling_team']
    )['total_runs']
    .sum()
    .reset_index()
)
```

```python
player_match = player_match.merge(
    matches, on='matchId', how='left'
)
```

```python
matches.groupby('season')['matchId'].count()
```

```python
player_match.groupby('batsman')['total_runs'].sum() \
             .sort_values(ascending=False).head(10)
```

```python
player_match.groupby('venue')['total_runs'].mean() \
             .sort_values(ascending=False).head(10)
```

```python
player_match = player_match.sort_values(['batsman','matchId'])
```

```python
player_match['career_avg'] = (
    player_match.groupby('batsman')['total_runs']
    .expanding().mean()
    .reset_index(level=0, drop=True)
)
```

```python
player_match['recent_form_5'] = (
    player_match.groupby('batsman')['total_runs']
    .rolling(5, min_periods=1).mean()
    .reset_index(level=0, drop=True)
)
```

```python
player_match['opponent_avg'] = (
    player_match.groupby(['batsman','bowling_team'])['total_runs']
    .transform('mean')
)
```

```python
player_match['venue_avg'] = (
    player_match.groupby(['batsman','venue'])['total_runs']
    .transform('mean')
)
```

```python
final_df = player_match.dropna()
```
