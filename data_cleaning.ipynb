```python
import pandas as pd
import numpy as np

# Loading raw IPL ball-by-ball dataset
df = pd.read_csv("data/deliveries_initial.csv")

# Inspecting structure and data types
df.info()
df.head()
```

```python
# Standardizing column names to snake_case and lowercase
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
)

df.columns
# Dropping redundant over_ball column
if "over_ball" in df.columns:
    df = df.drop(columns=["over_ball"])
# Normalizing text-based columns
text_cols = [
    "batting_team", "bowling_team",
    "batsman", "non_striker", "bowler",
    "dismissal_kind", "player_dismissed"
]

for col in text_cols:
    df[col] = (
        df[col]
        .astype(str)
        .str.strip()
        .str.lower()
    )
df.info()
df.head()
```

```python
# Mapping old team names to current franchise names
team_name_map = {
    "delhi daredevils": "delhi capitals",
    "deccan chargers": "sunrisers hyderabad",
    "kings xi punjab": "punjab kings",
    "rising pune supergiant": "rising pune supergiants",
    "rising pune supergiants": "rising pune supergiants",
    "pune warriors": "defunct_pune_warriors",
    "gujarat lions": "defunct_gujarat_lions",
    "kochi tuskers kerala": "defunct_kochi_tuskers"
}

df["batting_team"] = df["batting_team"].replace(team_name_map)
df["bowling_team"] = df["bowling_team"].replace(team_name_map)
df.info()
df.head()
```

```python
# Replacing NaNs in delivery-level boolean/numeric extras
extras_cols = ["iswide", "isnoball", "byes", "legbyes", "penalty"]
df[extras_cols] = df[extras_cols].fillna(0)

# Handling missing dismissal information explicitly
df["dismissal_kind"] = df["dismissal_kind"].replace("nan", "none")
df["player_dismissed"] = df["player_dismissed"].replace("nan", "none")
df.info()
df.head()
```

```python
# Computing total runs conceded on the delivery
df["total_runs"] = (
    df["batsman_runs"]
    + df["extras"]
    + df["byes"]
    + df["legbyes"]
    + df["penalty"]
)

# Creating striker-specific wicket indicator
df["is_wicket"] = np.where(
    (df["player_dismissed"] != "none") &
    (df["player_dismissed"] == df["batsman"]),
    1,
    0
)
df.info()
df.head()
```

```python
# Removing logically invalid deliveries
df = df[df["inning"].isin([1, 2])]
df = df[df["over"].between(0, 20)]
df = df[df["ball"].between(1, 6)]
df = df[df["batsman_runs"] >= 0]
df = df[df["total_runs"] >= 0]
df.info()
df.head()
```

```python
# Converting date column to datetime
df["date"] = pd.to_datetime(df["date"], errors="coerce")

# Removing rows with invalid dates
df = df[df["date"].notna()]

# Casting integer columns explicitly
int_cols = [
    "matchid", "inning", "over", "ball",
    "batsman_runs", "extras",
    "total_runs", "is_wicket"
]

df[int_cols] = df[int_cols].astype(int)
df.info()
df.head()
```

```python
# Sorting dataset chronologically for time-series safety
df = df.sort_values(
    by=["date", "matchid", "inning", "over", "ball"]
).reset_index(drop=True)

# Saving the cleaned dataset
df.to_csv("data/deliveries_cleaned.csv", index=False)

print("Section 5.1 data cleaning completed successfully.")
df.info()
df.head()
```
