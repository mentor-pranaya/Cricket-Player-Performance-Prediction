```python
import pandas as pd
import joblib

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
```

Load Feature Dataset 

```python
df = pd.read_csv("../data/processed/features.csv")

# Removing the Output column from the feature set
X = df.drop(columns=["target_next_runs"])
y = df["target_next_runs"]

X.head()
```

Define Numerical Columns

```python
num_features = X.columns.tolist()
```

PreProcessing Pipeline

```python
num_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("num", num_pipeline, num_features)
    ]
)
```

Fit and save Pipeline

```python
preprocessor.fit(X)

joblib.dump(preprocessor, "../models/feature_pipeline.pkl")

print("✅ Preprocessing pipeline saved as feature_pipeline.pkl")
```
