```python
import pandas as pd
import numpy as np
from google.colab import drive

# Mount Drive
drive.mount('/content/drive')

# Define Paths
base_path = '/content/drive/MyDrive/IPL_Performance_Prediction/data/'
raw_path = base_path + 'raw/'
processed_path = base_path + 'processed/'

# Load RAW Data (The original Kaggle files)
matches = pd.read_csv(raw_path + 'matches.csv')
deliveries = pd.read_csv(raw_path + 'deliveries.csv')

print("Raw Data Loaded!")
print(f"Matches: {matches.shape}")
print(f"Deliveries: {deliveries.shape}")
```

```python
# --- 1. Fix Team Names (IPL History) ---
# Teams often change names. We map old names to current ones.
team_mapping = {
    'Delhi Daredevils': 'Delhi Capitals',
    'Rising Pune Supergiant': 'Rising Pune Supergiants',
    'Kings XI Punjab': 'Punjab Kings'
}

# Apply the mapping
matches.replace(team_mapping, inplace=True)
deliveries.replace(team_mapping, inplace=True)

print("Team names standardized.")

# --- 2. Standardize Column Names ---
matches.rename(columns={'id': 'match_id', 'season': 'year'}, inplace=True)
deliveries.rename(columns={'match_id': 'match_id', 'batsman': 'batsman', 'bowler': 'bowler'}, inplace=True)

# --- 3. Handle Missing Values (Cities) ---
# Fill missing cities based on venue if possible, or drop
matches['city'].fillna(matches['venue'].apply(lambda x: x.split(' ')[0]))

print("Columns renamed and missing values handled.")
```

```python
# Save to 'processed' folder
matches.to_csv(processed_path + 'matches_cleaned.csv', index=False)
deliveries.to_csv(processed_path + 'deliveries_cleaned.csv', index=False)

print(f"Clean data saved to: {processed_path}")
```
