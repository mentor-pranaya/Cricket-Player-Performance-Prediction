# *Cheching Duplicates*

```python
def check_duplicates(df, name):
    dup_count = df.duplicated().sum()
    if dup_count == 0:
        print(f" {name}: No duplicate rows found")
    else:
        print(f" {name}: {dup_count} duplicate rows found")
    return dup_count
```

## # *DATA STANDARDIZATION*

```python
team_id_map = dict(zip(teams_data['team_id'], teams_data['team_name']))
```

```python
team_alias_map = dict(zip(team_alias['team_id'], team_alias['alias_name']))
```

```python
def map_team_name(value):

    if value in team_id_map:
        return team_id_map[value]

    if value in team_alias_map:
        return team_alias_map[value]
    
    return value
```

```python
team_columns_matches = ['team1', 'team2', 'toss_winner', 'match_winner']

for col in team_columns_matches:
    if col in matches.columns:
        matches[col] = matches[col].apply(map_team_name)
```

```python
team_columns_ball = ['team_batting', 'team_bowling']

for col in team_columns_ball:
    if col in ball.columns:
        ball[col] = ball[col].apply(map_team_name)
```

# ID to Player Name

```python
player_id_to_name = dict(
    zip(players['player_id'], players['player_name'])
)
```

```python
def map_player_of_match(value):
    # Missing value
    if pd.isna(value):
        return "Not Awarded"
    
    #  ID exists in players table
    if value in player_id_to_name:
        return player_id_to_name[value]
    
    return value
```

```python
matches['player_of_match'] = matches['player_of_match'].apply(map_player_of_match)
```

# Imputation 

```python
def null_report(df, name):
    
    print(f"\n===== {name} =====")
    total_rows = df.shape[0]
    
    null_count = df.isnull().sum()
    null_percent = (null_count / total_rows) * 100
    
    report = pd.DataFrame({
        "Null_Count": null_count,
        "Null_Percent": null_percent
    }).sort_values(by="Null_Percent", ascending=False)
    
    report = report[report["Null_Count"] > 0]
    
    if report.empty:
        print("No null values found ")
    else:
        print(report)
```

```python
null_report(ball, "Ball By Ball Data")
null_report(matches, "Matches Data")
```

```python
ball['wicket_kind'] = ball['wicket_kind'].fillna('No Wicket')
ball['player_out'] = ball['player_out'].fillna('Not Out')
ball['fielders_involved'] = ball['fielders_involved'].fillna('None')
```

```python
matches['win_by_runs'] = matches['win_by_runs'].fillna(0)
matches['win_by_wickets'] = matches['win_by_wickets'].fillna(0)
```

```python
matches['city'] = matches['city'].fillna('Not Recorded')
matches['match_number'] = matches['match_number'].fillna('Unknown Match')
```

```python
null_report(ball, "Ball By Ball")
null_report(matches, "Matches")
```

# Outliers

```python
def outlier_stats_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower) | (df[column] > upper)]

    total_rows = df.shape[0]
    outlier_count = outliers.shape[0]
    outlier_percent = (outlier_count / total_rows) * 100

    return outlier_count, outlier_percent
```

```python
count, percent = outlier_stats_iqr(ball, "batter_runs")

print("Outliers in total_runs:", count)
print("Outlier percentage:", round(percent, 2), "%")
```

```python
numeric_cols_ball = [
    "batter_runs",
    "extras",
    "total_runs",
    "wide_ball_runs",
    "no_ball_runs",
    "leg_bye_runs",
    "bye_runs"
]

outlier_report = []

for col in numeric_cols_ball:
    count, percent = outlier_stats_iqr(ball, col)
    outlier_report.append([col, count, round(percent, 2)])

outlier_df = pd.DataFrame(
    outlier_report,
    columns=["Column", "Outlier_Count", "Outlier_Percentage"]
)

outlier_df
```
