```python
city_freq = batsman["city"].value_counts(normalize=True)

batsman["city_enc"] = batsman["city"].map(city_freq)
batsman.drop(columns=["city"], inplace=True)


city_freq = bowler["city"].value_counts(normalize=True)

bowler["city_enc"] = bowler["city"].map(city_freq)
bowler.drop(columns=["city"], inplace=True)
```

```python
batsman["season"] = batsman["season"].str.split("/").str[0].astype(int)
bowler["season"] = bowler["season"].str.split("/").str[0].astype(int)
```

```python
batsman["date"] = pd.to_datetime(batsman["date"])
batsman["match_year"]  = batsman["date"].dt.year
batsman["match_month"] = batsman["date"].dt.month

bowler["date"] = pd.to_datetime(bowler["date"])
bowler["match_year"]  = bowler["date"].dt.year
bowler["match_month"] = bowler["date"].dt.month
```
