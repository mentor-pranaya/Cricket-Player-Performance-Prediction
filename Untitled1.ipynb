```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("IPL_ball_by_ball_updated.csv")

# -----------------------------
# 1. Basic Shape Information
# -----------------------------
print("DATASET SHAPE")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("-" * 40)
```

```python

# -----------------------------
# 2. Column Information
# -----------------------------
print("COLUMN NAMES")
print(df.columns.tolist())
print("-" * 40)

print("COLUMN DATA TYPES")
print(df.dtypes)
print("-" * 40)
```

```python
# -----------------------------
# 3. Memory Usage
# -----------------------------
print("MEMORY USAGE")
print(df.memory_usage(deep=True).sum() / 1024**2, "MB")
print("-" * 40)
```

```python

# -----------------------------
# 4. Missing Values
# -----------------------------
print("MISSING VALUES (Count & %)")
missing_df = pd.DataFrame({
    "Missing Count": df.isna().sum(),
    "Missing %": (df.isna().sum() / len(df)) * 100
})
print(missing_df[missing_df["Missing Count"] > 0])
print("-" * 40)
```

```python
# -----------------------------
# 5. Numerical Summary
# -----------------------------
print("NUMERICAL SUMMARY")
print(df.describe())
print("-" * 40)
```

```python
# -----------------------------
# 6. Categorical Summary
# -----------------------------
print("CATEGORICAL SUMMARY")
print(df.describe(include=["object"]))
print("-" * 40)
```

```python
# -----------------------------
# 7. Unique Value Counts
# -----------------------------
print("UNIQUE VALUES PER COLUMN")
unique_counts = df.nunique().sort_values(ascending=False)
print(unique_counts)
print("-" * 40)
```

```python


# -----------------------------
# 8. Key Cricket Metrics
# -----------------------------
print("CRICKET-SPECIFIC OVERVIEW")

overview = {
    "Total Matches": df["match_id"].nunique(),
    "Total Seasons": df["season"].nunique(),
    "Total Venues": df["venue"].nunique(),
    "Total Balls": len(df),
    "Total Batsmen": df["striker"].nunique(),
    "Total Bowlers": df["bowler"].nunique(),
}

for k, v in overview.items():
    print(f"{k}: {v}")

print("-" * 40)
```
