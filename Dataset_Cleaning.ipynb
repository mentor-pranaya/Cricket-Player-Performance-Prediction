```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/IPL.csv")
df.head()
df.info()
df.describe()
```

```python
df.duplicated().sum()
df = df.drop_duplicates()
```

```python
list(df.columns)
```

```python
df['next_batter'] = df['next_batter'].fillna("None")
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
int_cols = [
    'innings', 'over', 'ball', 'ball_no',
    'bat_pos', 'balls_faced', 'runs_batter',
    'runs_extras', 'runs_total', 'runs_bowler',
    'runs_not_boundary', 'team_runs', 'team_balls',
    'team_wicket', 'balls_per_over', 'overs',
    'batter_runs', 'batter_balls', 'bowler_wicket'
]

df[int_cols] = df[int_cols].astype(int)
```

```python
cat_cols = [
    'match_type', 'batting_team', 'bowling_team',
    'batter', 'bowler', 'non_striker', 'venue',
    'city', 'season', 'gender', 'team_type', 'wicket_kind',
    'player_out', 'fielders', 'extra_type'
]

df[cat_cols] = df[cat_cols].astype('category')
```

```python
bool_cols = ['valid_ball', 'striker_out']
df[bool_cols] = df[bool_cols].astype('int8')
```

```python
df.dtypes
```

```python
num_cols = [
    'innings', 'over', 'ball', 'ball_no', 'bat_pos',
    'balls_faced', 'runs_batter', 'runs_extras',
    'runs_total', 'runs_bowler', 'runs_not_boundary',
    'team_runs', 'team_balls', 'team_wicket',
    'balls_per_over', 'overs', 'batter_runs', 'batter_balls', 'bowler_wicket'
]

for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5*IQR
    upper = Q3 + 1.5*IQR

    # Cap outliers to limits (Winsorizing)
    df[col] = df[col].clip(lower, upper)
```
